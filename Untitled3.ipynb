### 2. Mapeamento de Áreas e Processamento dos Dados
Os cursos serão classificados em 4 áreas com base no nome do curso (ou códigos, se aplicável, mas usaremos palavras-chave comuns dos nomes dos cursos para flexibilidade nos diferentes anos):

1. **Ciências Exatas e Tecnologia**: Engenharia, Computação, Física, Matemática, Química (tecnológica), Sistemas, Tecnologia, etc.
2. **Ciências da Natureza e Saúde**: Medicina, Enfermagem, Biologia, Farmácia, Nutrição, Odontologia, Veterinária, etc.
3. **Gestão e Negócios**: Administração, Contabilidade, Economia, Logística, Marketing, Turismo, etc.
4. **Ciências Humanas, Sociais e Artes**: Direito, Pedagogia, Psicologia, História, Letras, Artes, Ciências Sociais, Jornalismo, etc.

```python
import os
import pandas as pd

def classificar_area(nome_curso):
    if not isinstance(nome_curso, str):
        return '4 - Ciências Humanas, Sociais e Artes'
    nome = nome_curso.lower()

    # 1 - Ciências Exatas e Tecnologia
    if any(x in nome for x in ['engenharia', 'computaç', 'física', 'matemática', 'sistemas', 'tecnologia', 'química', 'informática', 'análise e des', 'dados', 'jogos', 'redes', 'telecom']):
        return '1 - Ciências Exatas e Tecnologia'

    # 2 - Ciências da Natureza e Saúde
    elif any(x in nome for x in ['medicina', 'enfermagem', 'biologia', 'farmácia', 'nutrição', 'odontologia', 'veterinária', 'fisioterapia', 'educação física', 'saúde', 'biomedi', 'terapia ocupacional', 'fonoaudiologia']):
        return '2 - Ciências da Natureza e Saúde'

    # 3 - Gestão e Negócios
    elif any(x in nome for x in ['administra', 'contábe', 'economia', 'gestão', 'negócios', 'logística', 'marketing', 'finanças', 'recursos humanos', 'comércio', 'processos gerenciais']):
        return '3 - Gestão e Negócios'

    # 4 - Ciências Humanas, Sociais e Artes
    else:
        return '4 - Ciências Humanas, Sociais e Artes'

pasta_busca = '/content/drive/MyDrive/INEP'
dados_consolidados = []

if os.path.exists(pasta_busca):
    arquivos = [f for f in os.listdir(pasta_busca) if f.upper().endswith('.CSV') and 'CURSOS' in f.upper()]
    arquivos.sort()

    for file in arquivos:
        file_path = os.path.join(pasta_busca, file)
        print(f"Lendo arquivo: {file}")

        ano = ''.join(filter(str.isdigit, file))
        ano = ano[:4] if (ano and len(ano) >= 4) else "Desconhecido"

        try:
            # Ler 1 linha para detectar colunas
            df_temp = pd.read_csv(file_path, sep=';', encoding='latin1', nrows=1)
            df_temp.columns = [col.upper().strip() for col in df_temp.columns]

            col_curso = next((col for col in df_temp.columns if 'NO_CURSO' in col or 'NOME_CURSO' in col or 'NO_CINE_ROTULO' in col), None)
            col_trancada = next((col for col in df_temp.columns if 'QT_SIT_TRANCADA' in col or 'QT_TRANCADA' in col or 'QT_ALUNO_SIT_TRANCADA' in col), None)
            col_mat = next((col for col in df_temp.columns if 'QT_MAT' in col or 'QT_ALUNO_MAT' in col or 'QT_MATRICULADO' in col), None)

            if col_curso and col_trancada and col_mat:
                col_curso_orig = df_temp.columns[df_temp.columns.get_loc(col_curso)]
                col_trancada_orig = df_temp.columns[df_temp.columns.get_loc(col_trancada)]
                col_mat_orig = df_temp.columns[df_temp.columns.get_loc(col_mat)]

                # Ler somente as colunas necessárias
                df = pd.read_csv(file_path, sep=';', encoding='latin1', usecols=[col_curso_orig, col_trancada_orig, col_mat_orig], low_memory=False)
                df.columns = [col.upper().strip() for col in df.columns]

                df[col_trancada] = pd.to_numeric(df[col_trancada], errors='coerce').fillna(0)
                df[col_mat] = pd.to_numeric(df[col_mat], errors='coerce').fillna(0)

                df['Área'] = df[col_curso].apply(classificar_area)

                # Agrupar e somar ambos os valores por área
                agrupado = df.groupby('Área').agg({
                    col_trancada: 'sum',
                    col_mat: 'sum'
                }).reset_index()

                # Aplicar a fórmula da porcentagem: trancadas / (matriculadas + trancadas) * 100
                denom = agrupado[col_mat] + agrupado[col_trancada]
                agrupado['TAXA_TRANCAMENTO'] = (agrupado[col_trancada] / denom.replace(0, 1)) * 100
                agrupado['Ano'] = ano

                dados_consolidados.append(agrupado[['Área', 'TAXA_TRANCAMENTO', 'Ano']])
                print(f"-> Sucesso ao processar {file} ({ano}).")
            else:
                print(f"Aviso: Colunas necessárias não identificadas em {file}.")
        except Exception as e:
            print(f"Erro ao processar {file}: {e}")
else:
    print(f"Pasta não encontrada: {pasta_busca}")

if dados_consolidados:
    df_final = pd.concat(dados_consolidados, ignore_index=True)
    display(df_final.head())
else:
    print("Nenhum arquivo pôde ser consolidado de forma robusta.")
```

### 3. Geração do Gráfico Comparativo
Com os dados consolidados, vamos plotar um gráfico de colunas agrupadas por ano para as 4 áreas.

```python
if 'df_final' in locals() and not df_final.empty:
    # Pivotar para o gráfico
    df_pivot = df_final.pivot_table(
        index='Ano',
        columns='Área',
        values='TAXA_TRANCAMENTO',
        aggfunc='mean'
    ).fillna(0)

    df_pivot = df_pivot.sort_index()

    # Plotar o gráfico
    fig, ax = plt.subplots(figsize=(16, 8))
    df_pivot.plot(kind='bar', width=0.8, ax=ax, colormap='viridis')

    plt.title('Taxa de Matrículas Trancadas por Ano e Área de Conhecimento (INEP)', fontsize=16, fontweight='bold')
    plt.xlabel('Ano', fontsize=12)
    plt.ylabel('Taxa de Trancamento (%)', fontsize=12)
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.1f}%'.format(y)))
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.legend(title='Áreas', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()

    plt.show()
else:
    print("Não há dados disponíveis para gerar o gráfico.")
```

### 4. Gráfico Consolidado Geral (Todas as Áreas - Sem divisão por Ano)
Vamos agregar o total de matrículas e de trancamentos de todos os anos juntos para calcular a taxa de trancamento geral por Área de Conhecimento.

```python
if 'df_final' in locals() and not df_final.empty:
    # Usando a mesma lógica robusta anterior para consolidar as taxas de forma precisa
    pasta_busca = '/content/drive/MyDrive/INEP'

    if os.path.exists(pasta_busca):
        arquivos = [f for f in os.listdir(pasta_busca) if f.upper().endswith('.CSV') and 'CURSOS' in f.upper()]

        total_trancadas_global = {}
        total_matriculadas_global = {}

        for file in arquivos:
            file_path = os.path.join(pasta_busca, file)
            try:
                df_temp = pd.read_csv(file_path, sep=';', encoding='latin1', nrows=1)
                df_temp.columns = [col.upper().strip() for col in df_temp.columns]

                col_curso = next((col for col in df_temp.columns if 'NO_CURSO' in col or 'NOME_CURSO' in col or 'NO_CINE_ROTULO' in col), None)
                col_trancada = next((col for col in df_temp.columns if 'QT_SIT_TRANCADA' in col or 'QT_TRANCADA' in col or 'QT_ALUNO_SIT_TRANCADA' in col), None)
                col_mat = next((col for col in df_temp.columns if 'QT_MAT' in col or 'QT_ALUNO_MAT' in col or 'QT_MATRICULADO' in col), None)

                if col_curso and col_trancada and col_mat:
                    df = pd.read_csv(file_path, sep=';', encoding='latin1', usecols=[col_curso, col_trancada, col_mat], low_memory=False)
                    df.columns = [col.upper().strip() for col in df.columns]

                    df[col_trancada] = pd.to_numeric(df[col_trancada], errors='coerce').fillna(0)
                    df[col_mat] = pd.to_numeric(df[col_mat], errors='coerce').fillna(0)
                    df['Área'] = df[col_curso].apply(classificar_area)

                    for area, grp in df.groupby('Área'):
                        total_trancadas_global[area] = total_trancadas_global.get(area, 0) + grp[col_trancada].sum()
                        total_matriculadas_global[area] = total_matriculadas_global.get(area, 0) + grp[col_mat].sum()
            except Exception:
                pass

        # Criar o DataFrame final consolidado
        dados_grafico = []
        for area in total_trancadas_global.keys():
            t_tranc = total_trancadas_global[area]
            t_mat = total_matriculadas_global[area]
            taxa_global = (t_tranc / (t_mat + t_tranc)) * 100
            dados_grafico.append({'Área': area, 'Taxa Geral (%)': taxa_global})

        # Ordenar de forma crescente
        df_geral = pd.DataFrame(dados_grafico).sort_values('Taxa Geral (%)', ascending=True)

        # Definir as cores destacando apenas "Ciências Exatas e Tecnologia"
        # Usamos uma cor viva (ex: coral/vermelho) para exatas e cinza neutro para os outros
        destaque_cor = '#FF6347'  # Tomato/Coral
        cor_neutra = '#C0C0C0'    # Silver/Cinza

        cores = [
            destaque_cor if "Exatas" in area else cor_neutra
            for area in df_geral['Área']
        ]

        # Plotar gráfico de barras simples de área única em ordem crescente
        plt.figure(figsize=(12, 6))
        bars = plt.bar(df_geral['Área'], df_geral['Taxa Geral (%)'], color=cores, width=0.6, edgecolor='black', linewidth=0.7)

        # Adicionar as porcentagens acima das barras
        for bar in bars:
            yval = bar.get_height()
            plt.text(bar.get_x() + bar.get_width()/2, yval + 0.3, f'{yval:.2f}%', ha='center', va='bottom', fontweight='bold', fontsize=11)

        plt.title('Porcentagem de pessoas que trancaram cursos nos últimos 16 anos (INEP)', fontsize=14, fontweight='bold')
        plt.xlabel('Área de Conhecimento', fontsize=12)
        plt.ylabel('Taxa de Trancamento (%)', fontsize=12)
        plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.1f}%'.format(y)))
        plt.ylim(0, max(df_geral['Taxa Geral (%)']) * 1.15)
        plt.grid(axis='y', linestyle='--', alpha=0.5)
        plt.tight_layout()
        plt.show()
    else:
        print("Caminho não encontrado para consolidar os dados gerais.")
else:
    print("Não há dados disponíveis para gerar o gráfico.")
```
